```python
data=([(1,'Virat'),(2,'Rohit')])
schema="id int, name string"

df=spark.createDataFrame(data,schema)
display(df)
```

```python
df.write.mode("overwrite").saveAsTable("acc_dev.karthik.cricket")
```

```python
%sql
select * from acc_dev.karthik.cricket;
```

```python
data=([(5,'sky','All Rounder'),(6,'Shubnam','Batsman')])
schema="id int, name string, role string"

df=spark.createDataFrame(data,schema)
display(df)
```
