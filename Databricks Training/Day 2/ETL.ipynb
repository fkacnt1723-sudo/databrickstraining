```python
from pyspark.sql.functions import *
```

## Reading and Extracting

```python
df=spark.read.csv("/Volumes/acc_dev/naval/raw/financial_dataset.csv",header=True,inferSchema=True)
```

```python
df.display()
```

## Transformation

```python
df.select("CustomerID", "TransactionType")
```

```python
df.filter("transactionType = 'Withdrawal'").display()
```

```python
df.select("customerID".alias("cust_id"))
```

```python
df.select(col("customerID").alias("cust_id"))
```

```python
df.select("CustomerID",col("TransactionID"),df.AccountType,df["AmountUSD"]).display()
```

```python
df.withColumnsRenamed({"customerID": "cust_id","transactionID": "trans_id"}).display()
```

```python
df.withColumn("Branch",upper(df.Branch)).display()
```

```python
df.withColumn("ingestion_date",current_timestamp()).display()
```
