## Task Solution

```python
from pyspark.sql.functions import *
```

```python
input_path = "/Volumes/acc_dev/naval/raw/financial_dataset.csv"
```

```python
df_task = spark.read.csv(input_path, header=True, inferSchema=True)
```

```python
%skip
df_task.columns
```

```python
newcols = ['transaction_id',
 'customer_id',
 'account_type',
 'transaction_type',
 'amount_usd',
 'branch',
 'transaction_date']
```

```python
df1 = df_task.toDF(*newcols).withColumn("ingested_date",current_date()).withColumn("path",col("_metadata.file_name"))
```

```python
df1.write.mode("overwrite").saveAsTable("acc_dev.karthik.financials")
```
